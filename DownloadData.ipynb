# Downlaod Data from Edmond

File IDs are pinned on purpose. EDMOND serves newer versions of `Bulk`, `GBs`, `Point`, `Surfaces` and `Volume`, but those break `Verification.ipynb`: the new `Bulk.tgz` has no `Bulk.csv`, the `structure_file` paths in the new `Volume.csv` don't resolve, and the new `Surfaces.csv`/`GBs.csv` are missing most `structure_model_file` entries.

`SQS.tgz` holds the SQS test set (603 structures, results for all 8 MTPs).

```python
!mkdir -p Data
!wget https://edmond.mpg.de/api/access/datafile/252770 -O Data/Bulk.tgz
!wget https://edmond.mpg.de/api/access/datafile/252803 -O Data/Point.tgz
!wget https://edmond.mpg.de/api/access/datafile/252771 -O Data/GBs.tgz
!wget https://edmond.mpg.de/api/access/datafile/252772 -O Data/PhaseDiagram.tgz
!wget https://edmond.mpg.de/api/access/datafile/252773 -O Data/Surfaces.tgz
!wget https://edmond.mpg.de/api/access/datafile/252774 -O Data/TrainingData.tgz
!wget https://edmond.mpg.de/api/access/datafile/252776 -O Data/Volume.tgz
!wget https://edmond.mpg.de/api/access/datafile/334199 -O Data/SQS.tgz
```
